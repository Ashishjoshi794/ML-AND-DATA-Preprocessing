```python
import seaborn as sns
```

```python
df = sns.load_dataset('taxis')
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler # importing MinMaxScaler from sklearn.preprocessing 
```

```python
min_max = MinMaxScaler()
```

```python
min_max
```

```python
df[['distance', 'fare', 'tip', 'total']]
```

```python
max(df['distance']), min(df['distance']), max(df['fare']), min(df['fare']), max(df['tip']), min(df['tip']), max(df['total']), min(df['total']) # checking the max and min values of the columns
```

```python
max(df['total']), min(df['total'])
```

```python
min_max.fit(df[['distance', 'fare', 'tip', 'total']]) # fitting the data to the MinMaxScaler
```

```python
min_max.transform(df[['distance', 'fare', 'tip', 'total']]) # transforming the data to the MinMaxScaler
```

```python
min_max.fit_transform(df[['distance', 'fare', 'tip', 'total']]) # fitting and transforming the data to the MinMaxScaler
```

```python
import pandas as pd
df2 = pd.DataFrame(min_max.fit_transform(df[['distance', 'fare', 'tip', 'total']]), columns=['distance', 'fare', 'tip', 'total']) # creating a new dataframe with the transformed data
```
