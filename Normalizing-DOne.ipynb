```python
import numpy as np
data = np.array([[1, 2, 3],
                 [4, 5, 6]])

normalized_data = normalize(data, norm='l2')
print(normalized_data)
```

In Simple Terms

normalize() → Makes each row have length 1

MinMaxScaler() → Makes each column fit between 0 and 1

StandardScaler() → Makes each column have mean 0, std 1

| Method               | What It Does                                               | Works On          | Typical Range                  | Purpose                                                 |
| -------------------- | ---------------------------------------------------------- | ----------------- | ------------------------------ | ------------------------------------------------------- |
| **`normalize()`**    | Scales each row (sample) to unit norm                      | Each row (sample) | Depends on norm (L1, L2, etc.) | Useful for text data, distance-based models             |
| **`MinMaxScaler`**   | Scales each feature (column) to a range (default 0–1)      | Each feature      | [0, 1] (or custom)             | Useful for neural networks, feature scaling             |
| **`StandardScaler`** | Centers features around 0 and scales by standard deviation | Each feature      | ~[-3, 3]                       | Useful for models sensitive to variance (e.g. SVM, PCA) |
